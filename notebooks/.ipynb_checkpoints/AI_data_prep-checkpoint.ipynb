```python
from pyspark.sql.functions import col

# List of relevant columns based on provided parameters
relevant_columns = [
    'High', 'Low', 'Close', 'Volume', 'Open',
    'Timestamp','weekday_num','month_num','is_weekly_expiration','is_monthly_expiration',
    'SMA', 'EMA', 'RSI', 
    'MACD', 'MACD_signal', 
    'BBANDS_upper', 'BBANDS_middle','BBANDS_lower'
]
# Subset the DataFrame
BNF_df_sub = BNF_df.select([col(c) for c in relevant_columns])

# List of columns to check for null values
na_columns_to_check = [
    'SMA', 'EMA', 'RSI', 
    'MACD', 'MACD_signal', 
    'BBANDS_upper', 'BBANDS_middle','BBANDS_lower'
]

# Remove rows with null values in the specified columns
BNF_df_cleaned = BNF_df_sub.dropna(subset=na_columns_to_check)

print("Subset process complete..")
# BNF_df_cleaned.printSchema()
```

```python
# Get a summary of the DataFrame
summary = BNF_df_sub.describe()

# Show the summary
summary.show()

# Calculate the count of null values for each column
null_counts = BNF_df_sub.select([F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in BNF_df.columns])

# Show the null counts
null_counts.show()
```
