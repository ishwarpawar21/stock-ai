```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

def preprocess_data(train_df: pd.DataFrame, test_df: pd.DataFrame):
    # Handle missing values (if any)
    train_df = train_df.fillna(method='ffill')
    test_df = test_df.fillna(method='ffill')
    
    # Convert 'Timestamp' to datetime and extract additional features if needed
    train_df['Timestamp'] = pd.to_datetime(train_df['Timestamp'])
    test_df['Timestamp'] = pd.to_datetime(test_df['Timestamp'])
    train_df['weekday_num'] = train_df['Timestamp'].dt.weekday
    test_df['weekday_num'] = test_df['Timestamp'].dt.weekday
    
    # Define feature columns and target columns
    feature_columns = ['High', 'Low', 'Open','Close', 'weekday_num','EMA','RSI','SMA']
    target_columns = ['High_next', 'Low_next']

    # Normalize the features
    scaler = MinMaxScaler()
    x_train_scaled = scaler.fit_transform(train_df[feature_columns])
    x_test_scaled = scaler.transform(test_df[feature_columns])

    # Targets
    y_train = train_df[target_columns].values
    y_test = test_df[target_columns].values

    # Reshape input data to 3D for LSTM
    time_step = 60
    x_train_reshaped = np.array([x_train_scaled[i:i+time_step] for i in range(len(x_train_scaled) - time_step)])
    x_test_reshaped = np.array([x_test_scaled[i:i+time_step] for i in range(len(x_test_scaled) - time_step)])
    y_train = y_train[time_step:]
    y_test = y_test[time_step:]

    return x_train_reshaped, y_train, x_test_reshaped, y_test, scaler


```
