```python
# Extract data from the parsed JSON object
status = response_object.get("status")
candles = response_object.get("data", {}).get("candles", [])
# Convert the candles data to the correct types
candles = [
        [
            str(candle[0]),
            float(candle[1]),
            float(candle[2]),
            float(candle[3]),
            float(candle[4]),
            int(candle[5]),
            int(candle[6])
        ]
        for candle in candles
    ]
print("converted ...")
```

```python
# Create a Spark session
spark = SparkSession.builder.appName("Process JSON to Parquet").getOrCreate()
    
# Define schema
schema = StructType([
            StructField("Timestamp", StringType(), True),
            StructField("Open", DoubleType(), True),
            StructField("High", DoubleType(), True),
            StructField("Low", DoubleType(), True),
            StructField("Close", DoubleType(), True),
            StructField("Volume", LongType(), True),
            StructField("Open Interest", LongType(), True)])
# Create Spark DataFrame from the list of candles with specified schema
spark_df = spark.createDataFrame(candles, schema=schema)
print(spark_df.show(5))
```
